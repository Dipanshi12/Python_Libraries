# Importing Matplotlib

```python
import matplotlib.pyplot as plt
```

# Importing Pandas

```python
import pandas as pd
```

# Histogram

```python
# read csv file
df = pd.read_csv("Air.csv")
```

```python
df.columns
```

```python
# Creating Histogram
plt.hist(df['Total Landed Weight'],bins=10)
plt.show()
```

```python
# Creating Histogram with labels and title
plt.hist(df['Total Landed Weight'],bins=10)
plt.xlabel("Landed Weight")
plt.ylabel("Frequency")
plt.title("Distribution of landed weight")
plt.legend("Distribution")
plt.show()
```

```python
# Creating Histogram with labels and title and other changes
plt.hist(df['Total Landed Weight'],bins=10)
plt.xlabel("Landed Weight",color="r")
plt.ylabel("Frequency",color="r")
plt.title("Distribution of landed weight",color="b")
plt.legend("Distribution")
plt.show()
```

```python
# Edgecolor
plt.hist(df['Total Landed Weight'],bins=10,edgecolor="black")
plt.xlabel("Landed Weight",color="r")
plt.ylabel("Frequency",color="r")
plt.title("Distribution of landed weight",color="b")
plt.legend("Distribution")
plt.show()
```

```python
# Transparency
plt.hist(df['Total Landed Weight'],bins=10,edgecolor="black",alpha=0.5)
plt.xlabel("Landed Weight",color="r")
plt.ylabel("Frequency",color="r")
plt.title("Distribution of landed weight",color="b")
plt.legend("Distribution")
plt.show()
```

```python
# Orientation
plt.hist(df['Total Landed Weight'],bins=10,edgecolor="black",orientation="horizontal")
plt.xlabel("Frequency",color="r")
plt.ylabel("Landed Weight",color="r")
plt.title("Distribution of landed weight",color="b")
plt.legend("Distribution")
plt.show()
```

```python
# Grid
plt.hist(df['Total Landed Weight'],bins=10,edgecolor="black")
plt.xlabel("Frequency",color="r")
plt.ylabel("Landed Weight",color="r")
plt.title("Distribution of landed weight",color="b")
plt.legend("Distribution")

plt.grid()
plt.show()
```

```python
# Comparison
plt.hist(
    df["Total Landed Weight"].astype(float),
    bins=10,
    alpha=0.5,
    label="Landed Weight"
)

plt.hist(
    df['Landing Count'].astype(float),
    bins=10,
    alpha=0.5,
    label='Landing Count'
)

plt.legend()
plt.show()
```
