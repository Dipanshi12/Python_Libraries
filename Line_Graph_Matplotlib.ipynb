# Matplotlib

# Importing library

```python
import matplotlib.pyplot as plt
```

# Basic Graph

```python
x =[10,20,30,40,50]
y =[2,6,9,4,1]
plt.plot(x,y)
plt.show()
```

# Marker

```python
plt.plot(x,y,marker = "o")
plt.show()
```

```python
plt.plot(x,y,marker = "*")
plt.show()
```

# markersize

```python
plt.plot(x,y,marker = "*",markersize = "5")
plt.show()
```

# colour

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red")
plt.show()
```

# Title

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red")
plt.title("Line Graph")
plt.show()
```

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red")
plt.title("Line Graph",color="Blue")
plt.show()
```

# Label

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red")
plt.title("Line Graph",color="b")
plt.xlabel("Worker",color="y")
plt.ylabel("Time",color="g")
plt.show()
```

# Linestyle

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red",linestyle="dashdot")
plt.title("Line Graph",color="b")
plt.xlabel("Worker",color="y")
plt.ylabel("Time",color="g")
plt.show()
```

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red",linestyle="dashed")
plt.title("Line Graph",color="b")
plt.xlabel("Worker",color="y")
plt.ylabel("Time",color="g")
plt.show()
```

```python
plt.plot(x,y,marker = "o",markersize ="5",color = "red",linestyle="dotted")
plt.title("Line Graph",color="b")
plt.xlabel("Worker",color="y")
plt.ylabel("Time",color="g")
plt.show()
```

# Other way

```python
plt.plot(x,y,"r*--")
plt.title("Line Graph",color="b")
plt.xlabel("Worker",color="y")
plt.ylabel("Time",color="g")
plt.show()
```
